##### Tuples can be defined in various ways

```python
t1 = 1, 2, 3, 4, 5  

print(t1)
type(t1)
```

```python
t2 = 5.9,

print(t2)
type(t2)
```

```python
len(t2)
```

```python
# Single value tuple

t3 = 2,

print(t3)
type(t3)
```

```python
# This is not tuple

t4 = 1
t5 = (34)
```

```python
print(type(t4))
print(type(t5))
```

##### Concatination

```python
tup1 = ("Welcome", "to", "today's")
tup2 = ("class", "on", "Python" )

print(" ".join(tup1 + tup2))
```

```python
t1 = (1, 2, 3, 4, 5)
t2 = (43, 76, 89, 6)

t3 = t1 + t2

t3 = sorted(t3) # Here I sorted the tuple it will be automatically typecasted to list.

print(t3)

type(t3)
```

##### Immutability

```python
t = ("Hello!!", "Welcome" , 5, 10, "world", "of", "Analytics!!")
```

```python
t[3]
```

```python
t[3] = 5
```

```python
t[:2]
```

```python
t = ("Hello!!", "Welcome" , 5, 10, "world", "of", "Analytics!!")

t[-3:]
```

```python
# How to update the 2nd and 3rd element

new_tup = t[:2] + ("to", "the") + t[4:]

new_tup
```

```python
type(new_tup)
```

```python
t = (1, 2, 3)

new_t = t + "hello"
```

```python
new_t = t + ("hello",)

new_t
```

```python
new_t = t + 4
```

```python
new_t = t + (4,)

new_t
```

```python
t = ("disco", 12, 4.5)

t[0][2]
```

##### Sorting a tuple

```python
t = (2, 3, 4, 5, 8, 6, 9, 7)

new_t = sorted(t)

print(new_t)
type(new_t)
```

```python
# Typecasting

t2 = tuple(new_t)

print(t2)
```

```python
x = sorted(t)

tuple(x)
```

```python
# Nested Tuple

t = (1,5,"Hello","World", ("Java", "Python"))
```

```python
print(t[4][0])
print(t[4][-2])
print(t[-1][0])
print(t[-1][-2])
```
