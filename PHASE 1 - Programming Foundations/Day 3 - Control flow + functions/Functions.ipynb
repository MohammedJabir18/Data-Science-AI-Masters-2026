#### Functions

```python
def odd_even(n):
    if n % 2 != 0:
        print(f"{n} is a odd number")
    else:
        print(f"{n} is a even number")
```

```python
odd_even(5)
```

```python
odd_even(88)
```

```python
odd_even(21)
```

##### Factorial

```python
def fact_num(n):
    fact = 1
    
    for i in range(1, n+1):
        fact = fact * i

    return fact
```

```python
fact_num(4)
```

```python
fact_num(5)
```

```python
fact_num(6)
```

```python
# Recursion method

def fact_rec(n):
    return 1 if n <= 1 else n * fact_rec(n - 1)
```

```python
fact_rec(5)
```

```python
fact_rec(4)
```

```python
fact_rec(6)
```

##### Sum of all natural numbers till the number provided

```python
def sum_nat(n):
    num = 0

    for i in range(1, n+1):
        num = num + i
    
    return num
```

```python
sum_nat(4)
```

```python
sum_nat(5)
```

```python
def sum_nat_rec(n):
    return 1 if n <= 1 else n + sum_nat_rec(n - 1)
```

```python
sum_nat_rec(4)
```

```python
sum_nat_rec(5)
```

Create a lambda function that checks a number is even or odd

```python
x = lambda x: f"{x} is an Even number" if x % 2 == 0 else f"{x} is an Odd number"
```

```python
x(25)
```

```python
x(8)
```

```python
x(10)
```
