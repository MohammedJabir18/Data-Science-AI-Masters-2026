##### File Handling

Reading a file

```python
file = open("sample.txt", "r")

for line in file:
    print(line.strip())

file.close()
```

```python
file = open("sample.txt", "r")

print(file.read())

file.close()
```

```python
with open("sample.txt", "r") as file:
    print(file.read())
```

```python
with open("sample.txt", "r") as file:
    print(file.read(50))
    print(file.read(8))
```

```python
with open("sample.txt", "r") as file:
    print(file.readline())
```

##### Writing a file

```python
with open("sample.txt", "w") as f:
    f.write("Hello World\n")
    f.write("This is Python Programming section")
```

```python
with open("sample.txt", "r") as f:
    print(f.read())
```

```python
with open("sample.txt", "r") as f:
    f.write("Hello Wolrd")
```

##### Append operation

```python
with open("sample.txt", "a") as f:
    f.write("\nI am a Data Scientist")
```

```python
with open("sample.txt", "r") as f:
    print(f.read())
```

```python
with open("sample.txt", "a") as f:
    f.write("\nI am a Data Scientist")
    f.seek(0)
    print(f.read())
```

```python
with open("sample.txt", "a+") as f:
    f.write("\nI live in India")
    f.seek(0) # Rewind the cursor back to the start of the file
    print(f.read())
```
