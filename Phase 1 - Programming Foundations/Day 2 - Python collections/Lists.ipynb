##### The Challenge: "The Mirror Split & Zip"

You are given a list of exactly 6 elements. Your task is to transform it into a new list according to these rules:

1. Split the list perfectly in half.
2. Reverse only the second half.
3. Zip them back together so the elements alternate: [First element of 1st half, First element of 2nd half, Second element of 1st half, Second element of 2nd half...]

```python
original = [10, 20, 30, 40, 50, 60]

half_list = original[:len(original) // 2]

second_half = original[len(original) // 2:]

rev_sec = second_half[::-1]

fused_list = [half_list[0], rev_sec[0], half_list[1], rev_sec[1], half_list[2], rev_sec[2]]

print(fused_list)
```
