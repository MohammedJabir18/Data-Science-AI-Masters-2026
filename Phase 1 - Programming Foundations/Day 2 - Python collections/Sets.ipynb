#### Sets

* Sets are a type of collection like lists, tuples storing mixed data
* Sets are enclosed within curly brackets and elements are written as comma-separated
* Sets are unordered
* Sets does not allow duplicates

```python
list1 = [1,2,3,4,5,6,5,7,2,2,2,3]
```

```python
type(list1)
```

```python
len(list1)
```

```python
set_1 = set(list1)

set_1
```

```python
len(set_1)
```

```python
print(f"The number of duplicates removed from the list when converts to the set: {len(list1) - len(set_1)}") 
```

##### Set Operations

```python
A = {0, 2, 4, 6, 8}
B = {1, 2, 3, 4, 5}
C = {9, 7, 5, 3, 1}
```

```python
# Union

print(A | B)
print(B | A | C)
print(A.union(B, C))
print(B.union(A))
```

```python
print(len(A | B))
print(len(A | B | C))
```

```python
# Intersection

A = {0, 2, 4, 6, 8}
B = {1, 2, 3, 4, 5}
C = {9, 7, 5, 3, 1}

print(A & B)
print(B & A & C)
print(A.intersection(B))
print(B.intersection(A, C))
```

```python
print(len(A & B))
print(len(A & B & C))
```

```python
# Difference

A = {0, 2, 4, 6, 8}
B = {1, 2, 3, 4, 5}
C = {9, 7, 5, 3, 1}

print(A - B) 
print(B - A) 
print(A.difference(B, C))
```

```python
# Symmetric Difference

A = {0, 2, 4, 6, 8}
B = {1, 2, 3, 4, 5}
C = {9, 7, 5, 3, 1}

print(A ^ B)
print(B.symmetric_difference(A))
```

```python
print(A.symmetric_difference(B, C))
```
