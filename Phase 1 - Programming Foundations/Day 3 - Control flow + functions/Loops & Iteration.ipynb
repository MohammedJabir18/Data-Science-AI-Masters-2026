Loops

Check whether a number is even or odd

```python
x = int(input("Enter a number: "))
print(f"You are entered: {x}")

if x % 2 == 0:
    print(f"{x} is an even number")
else:
    print(f"{x} is an odd number")
```

```python
n = int(input("Enter your age: "))
print(f'Your age is {n}')

if n >= 18 and n <= 50:
    print(f"You are welcome")
else:
    print(f"OOps! You can't enter this party")
```

Program which offers various discounts on purchase bills

```python
x = int(input("Enter your bill amount: "))
print(f"{x} rupees bill amount purchased")

if x >= 1000:
    print("You have won 100$ voucher.")
elif x >= 500:
    print("You have won 50$ voucher.")
else:
    print("No Vouchers, Thank You")
```

##### Iteration

```python
# Iteration over a list of a intergers

list_1 = list(range(5, 15, 2))

print(list_1)
```

```python
for i in list_1:
    print(i)
```

```python
# Iterate over a String

str_1 = "Python"

for i in str_1:
    print(i)
```

```python
for num, i in enumerate(str_1, start=1):
    print(f"{num} character is: {i}")
```

```python
# Iterate over a dictionary

std_data = {
    1: ["Jabir", 25],
    2: ["Nabeel", 27],
    3: ["Shibu", 26],
    4: ["Varun", 24],
    5: ["Ajith", 28]
}

for i in std_data:
    print(i)
```

```python
for i, j in std_data.items():
    print(i, j)
```

```python
for i in std_data.keys():
    print(i)
```

```python
for i, j in std_data.values():
    print(i, j)
```

```python
# Iteration over a range values

for i in range(1, 6):
    print(i)
```

##### Comprehensions

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

list_2 = []

for i in list_1:
    list_2.append([i, len(i)])

print(list_2)
```

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

list_2 = [[i, len(i)] for i in list_1]

list_2
```

```python
lis = ["Hello", "python", "Welcome", "to", "Coding"]

d = {}

for i in lis:
    d.update({i: len(i)})

print(d)
```

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

d = {i: len(i) for i in list_1}

d
```
