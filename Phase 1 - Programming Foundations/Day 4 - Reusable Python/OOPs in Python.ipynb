#### Defining a class

* length and breadth as attributes
* init() - constructor of class
* self parameter - refers to the newly created instance of the class.
* attributes length and breadth are associated with self-keyword to identify them as instance variables

```python
class Rectangle:
    def __init__(self):
        self.length = 10
        self.breadth = 5
```

* create the object by calling name of the class followed by parenthesis.
* print the values using dot operator

![Constructor demonstration](../../assets/images/constructor_python.png)

A constructor is a special kind of method that Python calls when it instantiates an object using the definitions found in your class

```python
rect = Rectangle()

print(f"Length {rect.length}\nBreadth = {rect.breadth}") 
```

##### Parametrised Constructor

* parametrised constructor - dynamically assign the attribute values during object creation

```python
class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth
```

```python
rect = Rectangle(25, 30)

print(f"Length: {rect.length}\nBreadth: {rect.breadth}")
```

```python
# Instantiating a class
rect = Rectangle(5,10)

# calling the instance variable
rect.breadth
```

### iii. Adding a method to class

* calculate_area() - retutns the product of attributes length and breadth
* self - identifies its association with the instance

```python
class Rectangle:
    def __init__(self, length, breadth):
        self.leng = length
        self.bred = breadth

    def calcualte(self):
        return self.leng * self.bred
```

```python
# Instantiating the Rectangle class

rect = Rectangle(5, 10)

print(rect.leng)
print(rect.bred)
```

```python
rect.calcualte()
```

#### Significance of self:

* The attributes length and breadth are associated with an instance.
* Self makes sure that each instance refers to its own copy of attributes

```python
new_rect = Rectangle(15, 8)

print(f"Length: {new_rect.leng}\nBreadth: {new_rect.bred}\nArea: {new_rect.calcualte()}")
```

### iv. Inheritance and Overriding

```python
class Employee:
    def funct(self):
        print("Hello World")

class Department(Employee):
    pass


emp_dep = Department()
emp_dep.funct()
```

```python
class Shape:
    def set_color(self, color):
        self.color = color

    def calculate_area(self):
        pass

    def shape_color(self):
        print("Select a color from this: ", self.color)
        self.selected_color = input("Enter you want color: ").strip()
        return self.calculate_area() * self.color[self.selected_color]


class Circle(Shape):
    pi = 3.14

    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return Circle.pi * (self.radius ** 2)


color_price = {"Red": 10, "Blue": 6, "Green": 8, "Yellow": 2}

c = Circle(5)
c.set_color(color_price)
cost = c.shape_color()

print(f"Circle with radius: {c.radius}\nWhen colored: {c.selected_color}\nCost $: {cost}")
```
